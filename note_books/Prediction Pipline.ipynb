```python
import numpy as np
import pandas as pd
import re
import string
import pickle
```

```python
def remove_punctuation(text):
    for punctuation in string.punctuation:
        text=text.replace(punctuation,'')
    return text
```

```python
with open('../static/model/corpora/stopwords/english','r') as file:
    sw=file.read().splitlines()
```

```python
from nltk.stem import PorterStemmer
ps=PorterStemmer()
```

```python
def preprocessing(text):
    data=pd.DataFrame([text],columns=['tweet'])
    data["tweet"]= data["tweet"].apply(lambda x: " ".join(x.lower() for x in x.split()))
    data["tweet"]= data["tweet"].apply(lambda x: " ".join(re.sub(r'^https?:\/\/.*[\r\n]*','',x,flags=re.MULTILINE) for x in x.split()))
    data["tweet"] = data["tweet"].apply(remove_punctuation)
    data["tweet"] = data["tweet"].str.replace(r'\d+', '',regex=True)
    data["tweet"]= data["tweet"].apply(lambda x: " ".join(x for x in x.split() if x not in sw))
    data["tweet"]= data["tweet"].apply(lambda x: " ".join(ps.stem(x) for x in x.split()))

    return data["tweet"]
```

```python
vocab = pd.read_csv('../static/model/vocabulary.txt',header=None)
tokens = vocab[0].tolist()
```

```python
tokens
```

```python
def vectorizer(ds,vocabulary):
    vectorized_lst=[]

    for sentence in ds:
        sentence_lst=np.zeros(len(vocabulary))

        for i in range(len(vocabulary)):
            if vocabulary[i] in sentence.split():
                sentence_lst[i]=1

        vectorized_lst.append(sentence_lst)

    vectorized_lst_new=np.asarray(vectorized_lst,dtype=np.float32)

    return vectorized_lst_new
```

```python
with open('../static/model/model.pickle','rb') as f:
    model = pickle.load(f)
```

```python
def get_prediction(vectorized_text):
    prediction = model.predict(vectorized_text)
    if prediction ==1:
        return 'negative'
    else:
        return 'positive'
```

```python
txt1 = 'great product. i love it'
preprocessed_text1 = preprocessing(txt1)
vectorized_text1=vectorizer(preprocessed_text1,tokens)
prediction1=get_prediction(vectorized_text1)
prediction1
```

```python
txt2="bad product. I hate it"
preprocessed_text2=preprocessing(txt2)
vectorized_text2=vectorizer(preprocessed_text2,tokens)
prediction2=get_prediction(vectorized_text2)
prediction2
```
